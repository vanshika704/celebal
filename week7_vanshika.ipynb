```python
import os

from google.colab import files
from google.colab import userdata

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS

from langchain_google_genai import (
    GoogleGenerativeAIEmbeddings,
    ChatGoogleGenerativeAI,
)
```

```python
os.environ["GOOGLE_API_KEY"] = userdata.get("GOOGLE_API_KEY")
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)

uploaded = files.upload()

os.listdir("/content")

loader = PyPDFLoader("/content/🧠 WEEK-2_ CLASSICAL ML — FULL DEEP STRUCTURE.pdf")

documents = loader.load()

print("Total Pages:", len(documents))


documents[0]

print(documents[0].page_content)
print(documents[0].metadata)

for i in range(min(3, len(documents))):
    print("=" * 80)
    print(f"Page {i}")
    print("=" * 80)
    print(documents[i].page_content[:500])
    print("\n")
```

```python
# ==========================================
# STEP 1: Split the document into chunks
# ==========================================

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(documents)

print("Total Chunks:", len(chunks))

# Preview first chunk
print("\nFirst Chunk:\n")
print(chunks[0].page_content)

print("\nMetadata:")
print(chunks[0].metadata)


# ==========================================
# STEP 2: Create FAISS Vector Store
# ==========================================

vector_db = FAISS.from_documents(
    documents=chunks,
    embedding=embeddings
)

print(type(vector_db))


# ==========================================
# STEP 3: Create Retriever
# ==========================================

retriever = vector_db.as_retriever(
    search_kwargs={"k": 3}
)


# ==========================================
# STEP 4: Ask a Question
# ==========================================

query = "What is Linear Regression?"


# ==========================================
# STEP 5: Retrieve Relevant Chunks
# ==========================================

retrieved_docs = retriever.invoke(query)

print(f"\nRetrieved {len(retrieved_docs)} chunks.\n")


# ==========================================
# STEP 6: Display Retrieved Chunks
# ==========================================

for i, doc in enumerate(retrieved_docs, start=1):
    print("=" * 100)
    print(f"Retrieved Chunk {i}")
    print("=" * 100)
    print(doc.page_content)
    print("\n")


# ==========================================
# STEP 7: Create Context
# ==========================================

context = "\n\n".join(
    [doc.page_content for doc in retrieved_docs]
)


# ==========================================
# STEP 8: Create Prompt
# ==========================================

prompt = f"""
You are an AI assistant.

Answer the user's question ONLY using the context provided below.
If the answer is not present in the context, reply:
"I couldn't find the answer in the provided document."

Context:
{context}

Question:
{query}

Answer:
"""


# ==========================================
# STEP 9: Generate Answer
# ==========================================

response = llm.invoke(prompt)


# ==========================================
# STEP 10: Print Final Answer
# ==========================================

print("=" * 100)
print("FINAL ANSWER")
print("=" * 100)
print(response.content)
```
